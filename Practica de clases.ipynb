```python
class Vehiculo:
    def __init__(self, modelo, ano, precio):
        self.modelo = modelo
        self.ano = ano
        self.precio = precio
    
    def calcular_impuesto(self):
        impuesto = 0.30 * self.precio
        return impuesto

class Carro(Vehiculo):
    def __init__(self, modelo, ano, precio, marca):
        super().__init__(modelo, ano, precio)
        self.marca = marca
    
    def precio_descuento(self):
        descuento = 0.005 * self.precio
        return descuento

class Moto(Vehiculo):
    def __init__(self, modelo, ano, precio, cilindrada):
        super().__init__(modelo, ano, precio)
        self.cilindrada = cilindrada
        
    def calcular_impuesto(self):
        return 0.005 * self.precio
    
carro1 = Carro("Civic", 2020, 10000, "Honda")
moto1 = Moto("Africa Twin", 2023, 15000, 1100)

print("Este es un ", carro1.marca, " ", carro1.modelo, "fabricado en ", carro1.ano, ", y cuesta ", carro1.precio, "USD.\nSu impuesto es de ", carro1.calcular_impuesto(), " USD")

```
